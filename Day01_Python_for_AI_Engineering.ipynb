let's work with a realistic business dataset.

Imagine you're working as a Junior Data Analyst for an Australian retail company. Your manager gives you sales transaction records and asks you to prepare the data for analysis.

The dataset intentionally contains:
- A duplicate transaction.
- A missing discount value.
- Different products and locations.
- Sales quantities and unit prices.

```python
sales_data = {
    "order_id": [
        "O1001", "O1002", "O1003", "O1004",
        "O1005", "O1006", "O1007", "O1008",
        "O1009", "O1010", "O1003"
    ],

    "date": [
        "2026-01-03", "2026-01-04", "2026-01-05",
        "2026-01-06", "2026-01-07", "2026-01-08",
        "2026-01-09", "2026-01-10", "2026-01-11",
        "2026-01-12", "2026-01-05"
    ],

    "region": [
        "Brisbane", "Sydney", "Brisbane", "Melbourne",
        "Sydney", "Brisbane", "Melbourne", "Sydney",
        "Brisbane", "Melbourne", "Brisbane"
    ],

    "product": [
        "Laptop", "Monitor", "Keyboard", "Laptop",
        "Mouse", "Monitor", "Keyboard", "Laptop",
        "Mouse", "Monitor", "Keyboard"
    ],

    "quantity": [
        2, 3, 5, 1, 10, 2, 4, 1, 6, 2, 5
    ],

    "unit_price": [
        1200, 250, 45, 1200, 25, 250,
        45, 1200, 25, 250, 45
    ],

    "discount_pct": [
        0.10, 0.05, 0.00, 0.15, np.nan,
        0.05, 0.00, 0.10, 0.00, 0.10, 0.00
    ]
}

df = pd.DataFrame(sales_data)

print(df)
```

```python
# First five rows
print(df.head())

# Dataset dimensions
print("Shape:", df.shape)

# Column names and data types
df.info()

# Statistical summary
print(df.describe())
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate records:", duplicates)
```

```python
df = df.drop_duplicates()
```

```python
df["discount_pct"] = (
    df["discount_pct"].fillna(0)
)
```

```python
df["date"] = pd.to_datetime(df["date"])

df["quantity"] = df["quantity"].astype(int)

df["unit_price"] = df["unit_price"].astype(float)

print(df.dtypes)
```

```python
df["revenue"] = (
    df["quantity"]
    * df["unit_price"]
    * (1 - df["discount_pct"])
)

print(
    df[["order_id", "product", "revenue"]]
)
```

```python
revenue_array = np.array(df["revenue"])

print("Total revenue:", np.sum(revenue_array))
print("Average revenue:", np.mean(revenue_array))
print("Maximum revenue:", np.max(revenue_array))
print("Minimum revenue:", np.min(revenue_array))
```

```python
region_sales = (
    df.groupby("region")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(region_sales)
```

```python
product_sales = (
    df.groupby("product")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(product_sales)
```

```python
total_revenue = df["revenue"].sum()

best_region = region_sales.idxmax()

best_product = product_sales.idxmax()

average_order_value = df["revenue"].mean()

print("SALES PERFORMANCE REPORT")
print("------------------------")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Best Region: {best_region}")
print(f"Best Product: {best_product}")
```

first reusable data pipeline
Now let's combine everything into one reusable function.

This is an important programming concept for AI engineering: instead of writing the same cleaning operations repeatedly, we create a function that can process different datasets.

```python
def process_sales_data(data):

    # Create DataFrame
    data = pd.DataFrame(data).copy()

    # Remove duplicates
    data = data.drop_duplicates()

    # Handle missing values
    data["discount_pct"] = (
        data["discount_pct"].fillna(0)
    )

    # Convert date
    data["date"] = pd.to_datetime(data["date"])

    # Calculate revenue
    data["revenue"] = (
        data["quantity"]
        * data["unit_price"]
        * (1 - data["discount_pct"])
    )

    # Generate summary
    report = {
        "records_processed": len(data),
        "total_revenue": round(data["revenue"].sum(), 2),
        "average_revenue": round(data["revenue"].mean(), 2),
        "missing_values": int(data.isnull().sum().sum())
    }

    return data, report


cleaned_data, report = process_sales_data(sales_data)

print(report)
```

```python
cleaned_data.to_csv(
    "cleaned_sales_data.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```
